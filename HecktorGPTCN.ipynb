```python

# ============================================================
# HECKTOR GPT-2 — STRATIFIED 3-FOLD CROSS-VALIDATION
# Fully frozen GPT-2 backbone
# Original classifier architecture and hyperparameters
# All four original checkpoint-selection criteria
# Uses every patient with a valid HPV label
# Reports mean +/- sample SD across 3 held-out folds
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import sys
import gc
import random
import subprocess
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import StratifiedKFold, train_test_split

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    balanced_accuracy_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)

# Install gdown if needed
try:
    import gdown
except ImportError:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "gdown"
    ])
    import gdown

# Install transformers if needed
try:
    from transformers import GPT2Tokenizer, GPT2Model
except ImportError:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "transformers"
    ])
    from transformers import GPT2Tokenizer, GPT2Model


# ============================================================
# 2. GOOGLE DRIVE AND OUTPUT DIRECTORY — FIXED
# ============================================================

from pathlib import Path
import os

try:
    from google.colab import drive

    MOUNT_POINT = "/content/gdrive"

    os.makedirs(MOUNT_POINT, exist_ok=True)

    drive.mount(MOUNT_POINT)

    OUT_DIR = Path(
        "/content/gdrive/MyDrive/HECKTOR_GPT2_3Fold_results"
    )

except ImportError:

    OUT_DIR = Path("./HECKTOR_GPT2_3Fold_results")

OUT_DIR.mkdir(parents=True, exist_ok=True)

print("Output directory:", OUT_DIR)


# ============================================================
# 3. SETTINGS — SAME AS ORIGINAL MODEL
# ============================================================

SEED = 42

MODEL_NAME = "openai-community/gpt2"

N_FOLDS = 3

EPOCHS = 500
BATCH_SIZE = 64

LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
DROPOUT = 0.1

# Original split was 65/15/20.
# Within the 80% training portion, validation was 18.75%.
VALIDATION_SIZE = 0.1875

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)
print("Batch size:", BATCH_SIZE)
print("Epochs:", EPOCHS)
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Dropout:", DROPOUT)


def seed_everything(seed=SEED):

    os.environ["PYTHONHASHSEED"] = str(seed)

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.use_deterministic_algorithms(
        True,
        warn_only=True
    )

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything()


# ============================================================
# 4. DOWNLOAD AND LOAD HECKTOR DATASET
# ============================================================

DATA_FILE = "/content/HPV2025.xlsx"

GOOGLE_DRIVE_FILE_ID = "1DGDH2e6dEkpdwwtCvH6ds1CwJT2Hdmkv"

if not os.path.exists(DATA_FILE):

    print("\nDownloading HECKTOR dataset...")

    downloaded = gdown.download(
        id=GOOGLE_DRIVE_FILE_ID,
        output=DATA_FILE,
        quiet=False
    )

    if downloaded is None or not os.path.exists(DATA_FILE):
        raise FileNotFoundError(
            "Could not download HPV2025.xlsx. "
            "Check the Google Drive file availability."
        )

df = pd.read_excel(DATA_FILE)

print("\nDataset successfully loaded.")
print("Original dataset shape:", df.shape)

print("\nAvailable columns:")
print(df.columns.tolist())


# ============================================================
# 5. DEFINE FEATURES AND TARGET
# ============================================================

TARGET = "HPV Status"

FEATURES = [
    "Age",
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

missing = [
    col for col in [TARGET] + FEATURES
    if col not in df.columns
]

if missing:
    raise KeyError(
        f"Required Excel columns missing: {missing}"
    )


# ============================================================
# 6. NORMALISE HPV TARGET
# ============================================================

def normalize_target(x):

    if pd.isna(x):
        return np.nan

    s = str(x).strip().lower()

    if s in {
        "0", "0.0", "negative", "neg", "false", "no"
    }:
        return 0

    if s in {
        "1", "1.0", "positive", "pos", "true", "yes"
    }:
        return 1

    try:
        n = float(s)

        if n in (0, 1):
            return int(n)

    except (ValueError, TypeError):
        pass

    return np.nan


df["HPV_Label"] = df[TARGET].apply(normalize_target)

invalid_count = int(df["HPV_Label"].isna().sum())

df = df.dropna(subset=["HPV_Label"]).copy()

df["HPV_Label"] = df["HPV_Label"].astype(int)

df = df[FEATURES + ["HPV_Label"]].copy()

df = df.reset_index(drop=True)

if df.empty:
    raise ValueError("No valid HPV labels were found.")

class_counts = df["HPV_Label"].value_counts().reindex(
    [0, 1], fill_value=0
)

if (class_counts < N_FOLDS).any():
    raise ValueError(
        "Insufficient patients in one or more classes for 3-fold CV."
    )

print("\n======================================")
print("HECKTOR DATASET SUMMARY")
print("======================================")

print("Total original patients:", len(pd.read_excel(DATA_FILE)))
print("Patients dropped due to missing/invalid HPV label:", invalid_count)
print("Patients included in 3-fold CV:", len(df))

print("\nHPV class distribution:")
print(class_counts)

X = df[FEATURES].copy()

y = df["HPV_Label"].to_numpy(dtype=int)


# ============================================================
# 7. LOAD GPT-2 TOKENIZER
# ============================================================

tokenizer = GPT2Tokenizer.from_pretrained(MODEL_NAME)

tokenizer.pad_token = tokenizer.eos_token

print("\nGPT-2 tokenizer loaded.")


# ============================================================
# 8. ORIGINAL CLINICAL PROMPT CONSTRUCTION
# ============================================================

def display_value(x):

    if pd.isna(x):
        return "Unknown"

    if isinstance(x, (float, np.floating)) and float(x).is_integer():
        return str(int(x))

    return str(x).strip()


def format_stage(x, prefix):

    value = display_value(x)

    if value == "Unknown" or value.upper().startswith(prefix):
        return value

    try:
        n = float(value)

        if n.is_integer():
            return f"{prefix}{int(n)}"

    except ValueError:
        pass

    return f"{prefix}{value}"


def patient_to_text(row):

    age = display_value(row["Age"])
    gender = display_value(row["Gender"])

    tobacco = display_value(
        row["Tobacco Consumption"]
    )

    alcohol = display_value(
        row["Alcohol Consumption"]
    )

    performance = display_value(
        row["Performance Status"]
    )

    t_stage = format_stage(row["T-stage"], "T")
    n_stage = format_stage(row["N-stage"], "N")
    m_stage = format_stage(row["M-stage"], "M")

    return (
        f"This patient is a {age}-year-old {gender}. "
        f"Tobacco consumption is {tobacco}. "
        f"Alcohol consumption is {alcohol}. "
        f"Performance status is {performance}. "
        f"Tumour stage is {t_stage}. "
        f"Nodal stage is {n_stage}. "
        f"Metastatic stage is {m_stage}. "
        "Predict whether the patient is HPV positive or HPV negative."
    )


# ============================================================
# 9. FOLD-SPECIFIC IMPUTATION
# ============================================================

def impute_fold_data(train_df, val_df, test_df):

    train_df = train_df.copy()
    val_df = val_df.copy()
    test_df = test_df.copy()

    fill_values = {}

    for col in FEATURES:

        if pd.api.types.is_numeric_dtype(train_df[col]):

            value = train_df[col].median()

            if pd.isna(value):
                value = 0

        else:

            modes = (
                train_df[col]
                .dropna()
                .astype(str)
                .mode()
            )

            value = (
                modes.iloc[0]
                if not modes.empty
                else "Unknown"
            )

        fill_values[col] = value

        for split in (train_df, val_df, test_df):

            split[col] = split[col].fillna(value)

            if split[col].dtype == object:

                split[col] = (
                    split[col]
                    .astype(str)
                    .replace({
                        "": "Unknown",
                        "nan": "Unknown",
                        "None": "Unknown"
                    })
                )

    return train_df, val_df, test_df, fill_values


# ============================================================
# 10. DATASET CLASS
# ============================================================

class HPVTextDataset(Dataset):

    def __init__(self, texts, labels, max_length):

        self.items = []

        for text, label in zip(texts, labels):

            enc = tokenizer(
                text,
                max_length=max_length,
                padding="max_length",
                truncation=True,
                return_tensors="pt"
            )

            self.items.append((
                enc["input_ids"].squeeze(0),
                enc["attention_mask"].squeeze(0),
                int(label)
            ))

    def __len__(self):
        return len(self.items)

    def __getitem__(self, index):

        ids, mask, label = self.items[index]

        return {
            "input_ids": ids,
            "attention_mask": mask,
            "label": torch.tensor(
                label,
                dtype=torch.long
            )
        }


# ============================================================
# 11. GPT-2 CLASSIFIER
# ORIGINAL ARCHITECTURE — BACKBONE FULLY FROZEN
# ============================================================

class HPVNetGPT2Text(nn.Module):

    def __init__(self):

        super().__init__()

        self.gpt2 = GPT2Model.from_pretrained(MODEL_NAME)

        # Freeze every GPT-2 parameter.
        for param in self.gpt2.parameters():
            param.requires_grad = False

        self.gpt2.config.use_cache = False
        self.gpt2.config.pad_token_id = tokenizer.pad_token_id

        h = self.gpt2.config.hidden_size

        # Original classification head.
        self.classifier = nn.Sequential(
            nn.LayerNorm(h),
            nn.Dropout(DROPOUT),
            nn.Linear(h, 128),
            nn.GELU(),
            nn.Dropout(DROPOUT),
            nn.Linear(128, 2)
        )

    def forward(self, input_ids, attention_mask):

        with torch.no_grad():

            out = self.gpt2(
                input_ids=input_ids,
                attention_mask=attention_mask,
                use_cache=False,
                return_dict=True
            ).last_hidden_state

        positions = attention_mask.sum(dim=1) - 1

        pooled = out[
            torch.arange(out.size(0), device=out.device),
            positions
        ]

        return self.classifier(pooled)


# ============================================================
# 12. LOSS FUNCTION
# ============================================================

def make_weighted_loss(y_train):

    class_counts = np.bincount(
        y_train,
        minlength=2
    )

    class_weights = torch.tensor(
        len(y_train) / (2 * np.maximum(class_counts, 1)),
        dtype=torch.float32,
        device=device
    )

    def weighted_loss(logits, labels):

        losses = nn.functional.cross_entropy(
            logits,
            labels,
            reduction="none"
        )

        return (losses * class_weights[labels]).mean()

    return weighted_loss


# ============================================================
# 13. SCORING FUNCTIONS
# ============================================================

def get_scores(model, loader):

    model.eval()

    scores = []
    labels = []

    with torch.no_grad():

        for batch in loader:

            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)

            logits = model(ids, mask)

            scores.extend(
                torch.softmax(logits, dim=1)[:, 1]
                .cpu().numpy()
            )

            labels.extend(
                batch["label"].numpy()
            )

    return (
        np.asarray(scores),
        np.asarray(labels, dtype=int)
    )


def evaluate_loss(model, loader, loss_function):

    model.eval()

    total_loss = 0.0
    total_n = 0

    with torch.no_grad():

        for batch in loader:

            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            labels = batch["label"].to(device)

            loss = loss_function(
                model(ids, mask),
                labels
            )

            n = len(labels)

            total_loss += loss.item() * n
            total_n += n

    return total_loss / max(total_n, 1)


def choose_threshold(scores, labels):

    thresholds = np.unique(
        np.r_[0.0, scores, 1.0]
    )

    values = [
        (
            float(t),
            balanced_accuracy_score(
                labels,
                scores >= t
            )
        )
        for t in thresholds
    ]

    return max(
        values,
        key=lambda x: (x[1], -abs(x[0] - 0.5))
    )


# ============================================================
# 14. CHECKPOINT CRITERIA
# ============================================================

CHECKPOINT_NAMES = [
    "Validation_Loss",
    "Validation_ROC_AUC",
    "Validation_PR_AUC",
    "Validation_Balanced_Accuracy"
]

best_initial_values = {
    "Validation_Loss": float("inf"),
    "Validation_ROC_AUC": -float("inf"),
    "Validation_PR_AUC": -float("inf"),
    "Validation_Balanced_Accuracy": -float("inf")
}


# ============================================================
# 15. STRATIFIED 3-FOLD CROSS-VALIDATION
# ============================================================

skf = StratifiedKFold(
    n_splits=N_FOLDS,
    shuffle=True,
    random_state=SEED
)

fold_results = []
fold_histories = []

# Separate out-of-fold predictions for each checkpoint strategy.
oof_scores = {
    name: np.full(len(df), np.nan)
    for name in CHECKPOINT_NAMES
}

oof_predictions = {
    name: np.full(len(df), -1, dtype=int)
    for name in CHECKPOINT_NAMES
}

oof_fold = np.full(len(df), -1, dtype=int)

loader_args = {
    "batch_size": BATCH_SIZE,
    "num_workers": 0,
    "pin_memory": torch.cuda.is_available()
}


for fold, (train_val_idx, test_idx) in enumerate(
    skf.split(X, y),
    start=1
):

    print("\n" + "=" * 75)
    print(f"FOLD {fold}/{N_FOLDS}")
    print("=" * 75)

    seed_everything(SEED + fold)

    # --------------------------------------------------------
    # Outer training and held-out test partitions
    # --------------------------------------------------------

    fold_train_val = df.iloc[train_val_idx].copy()
    fold_test = df.iloc[test_idx].copy()

    # Original 65/15/20 proportions:
    # validation = 18.75% of the outer training portion.
    inner_train, inner_val = train_test_split(
        fold_train_val,
        test_size=VALIDATION_SIZE,
        random_state=SEED + fold,
        stratify=fold_train_val["HPV_Label"]
    )

    inner_train = inner_train.copy()
    inner_val = inner_val.copy()
    fold_test = fold_test.copy()

    # Fit imputation using training data only.
    inner_train, inner_val, fold_test, fill_values = (
        impute_fold_data(
            inner_train,
            inner_val,
            fold_test
        )
    )

    y_train = inner_train["HPV_Label"].to_numpy(dtype=int)
    y_val = inner_val["HPV_Label"].to_numpy(dtype=int)
    y_test = fold_test["HPV_Label"].to_numpy(dtype=int)

    print("\nTraining patients:", len(y_train))
    print("Training class counts:", np.bincount(y_train, minlength=2))

    print("\nValidation patients:", len(y_val))
    print("Validation class counts:", np.bincount(y_val, minlength=2))

    print("\nHeld-out test patients:", len(y_test))
    print("Test class counts:", np.bincount(y_test, minlength=2))

    if len(np.unique(y_val)) < 2 or len(np.unique(y_test)) < 2:
        raise ValueError(
            f"Fold {fold} validation or test set lacks one class."
        )

    # --------------------------------------------------------
    # Generate original prompts
    # --------------------------------------------------------

    train_texts = [
        patient_to_text(row)
        for _, row in inner_train.iterrows()
    ]

    val_texts = [
        patient_to_text(row)
        for _, row in inner_val.iterrows()
    ]

    test_texts = [
        patient_to_text(row)
        for _, row in fold_test.iterrows()
    ]

    # Original rule: determine MAX_LENGTH from training prompts.
    lengths = [
        len(tokenizer.encode(text, add_special_tokens=False))
        for text in train_texts
    ]

    MAX_LENGTH = min(
        1024,
        max(32, int(np.ceil(max(lengths) / 8) * 8))
    )

    print("\nMaximum training prompt length:", max(lengths))
    print("MAX_LENGTH:", MAX_LENGTH)

    # --------------------------------------------------------
    # Build datasets and loaders
    # --------------------------------------------------------

    train_ds = HPVTextDataset(
        train_texts,
        y_train,
        MAX_LENGTH
    )

    val_ds = HPVTextDataset(
        val_texts,
        y_val,
        MAX_LENGTH
    )

    test_ds = HPVTextDataset(
        test_texts,
        y_test,
        MAX_LENGTH
    )

    generator = torch.Generator().manual_seed(SEED + fold)

    train_loader = DataLoader(
        train_ds,
        shuffle=True,
        generator=generator,
        **loader_args
    )

    val_loader = DataLoader(
        val_ds,
        shuffle=False,
        **loader_args
    )

    test_loader = DataLoader(
        test_ds,
        shuffle=False,
        **loader_args
    )

    # --------------------------------------------------------
    # Initialise a fresh model for this fold
    # --------------------------------------------------------

    model = HPVNetGPT2Text().to(device)

    # Verify the GPT-2 backbone is completely frozen.
    assert all(
        not p.requires_grad
        for p in model.gpt2.parameters()
    )

    # Verify classifier head is trainable.
    assert all(
        p.requires_grad
        for p in model.classifier.parameters()
    )

    trainable_params = sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )

    total_params = sum(
        p.numel()
        for p in model.parameters()
    )

    print("\nTrainable parameters:", f"{trainable_params:,}")
    print("Total parameters:", f"{total_params:,}")
    print("GPT-2 backbone frozen: YES")
    print("Classifier head trainable: YES")

    # Optimise classifier head only.
    optimizer = torch.optim.AdamW(
        model.classifier.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    loss_function = make_weighted_loss(y_train)

    # --------------------------------------------------------
    # Initialise four independent best checkpoints
    # --------------------------------------------------------

    best = best_initial_values.copy()

    best_info = {}
    best_states = {}

    history = []

    # --------------------------------------------------------
    # TRAINING — ALL 500 EPOCHS
    # No early stopping
    # --------------------------------------------------------

    for epoch in range(1, EPOCHS + 1):

        model.train()

        # Match original training behaviour:
        # frozen GPT-2 remains in evaluation mode.
        model.gpt2.eval()

        loss_sum = 0.0
        n_sum = 0

        for batch in train_loader:

            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            labels = batch["label"].to(device)

            optimizer.zero_grad(set_to_none=True)

            logits = model(ids, mask)

            loss = loss_function(logits, labels)

            loss.backward()

            optimizer.step()

            n = len(labels)

            loss_sum += loss.item() * n
            n_sum += n

        train_loss = loss_sum / max(n_sum, 1)

        # Validation loss
        val_loss = evaluate_loss(
            model,
            val_loader,
            loss_function
        )

        # Validation probabilities
        val_scores, val_labels = get_scores(
            model,
            val_loader
        )

        val_auc = roc_auc_score(
            val_labels,
            val_scores
        )

        val_ap = average_precision_score(
            val_labels,
            val_scores
        )

        threshold, val_ba = choose_threshold(
            val_scores,
            val_labels
        )

        metrics = {
            "Validation_Loss": val_loss,
            "Validation_ROC_AUC": val_auc,
            "Validation_PR_AUC": val_ap,
            "Validation_Balanced_Accuracy": val_ba
        }

        history.append({
            "Fold": fold,
            "Epoch": epoch,
            "Train_Loss": train_loss,
            **metrics,
            "Validation_Threshold": threshold
        })

        # ----------------------------------------------------
        # Save best checkpoint for EACH original criterion
        # ----------------------------------------------------

        for name, value in metrics.items():

            if name == "Validation_Loss":
                improved = value < best[name]
            else:
                improved = value > best[name]

            if improved:

                best[name] = float(value)

                best_info[name] = {
                    "epoch": epoch,
                    "value": float(value),
                    "threshold": float(threshold)
                }

                # Store only classifier weights.
                # GPT-2 is frozen and reloaded from the same
                # pretrained checkpoint for every fold.
                best_states[name] = {
                    key: value.detach().cpu().clone()
                    for key, value in model.classifier.state_dict().items()
                }

        if epoch % 50 == 0:

            print(
                f"Fold {fold} | Epoch {epoch}/{EPOCHS} | "
                f"Train Loss={train_loss:.4f} | "
                f"Val Loss={val_loss:.4f} | "
                f"ROC-AUC={val_auc:.4f} | "
                f"PR-AUC={val_ap:.4f} | "
                f"BA={val_ba:.4f}"
            )

    fold_histories.extend(history)

    # --------------------------------------------------------
    # Evaluate all four best checkpoints on held-out fold
    # --------------------------------------------------------

    for name in CHECKPOINT_NAMES:

        model.classifier.load_state_dict(
            best_states[name]
        )

        model.eval()

        test_scores, test_labels = get_scores(
            model,
            test_loader
        )

        threshold = best_info[name]["threshold"]

        test_predictions = (
            test_scores >= threshold
        ).astype(int)

        test_auc = roc_auc_score(
            test_labels,
            test_scores
        )

        test_ap = average_precision_score(
            test_labels,
            test_scores
        )

        test_ba = balanced_accuracy_score(
            test_labels,
            test_predictions
        )

        test_f1 = f1_score(
            test_labels,
            test_predictions,
            zero_division=0
        )

        cm = confusion_matrix(
            test_labels,
            test_predictions,
            labels=[0, 1]
        )

        print("\n" + "-" * 65)
        print("Fold:", fold)
        print("Checkpoint:", name)
        print("Best epoch:", best_info[name]["epoch"])
        print("Best validation metric:", best_info[name]["value"])
        print("Validation threshold:", threshold)

        print("\nClassification report:")

        print(
            classification_report(
                test_labels,
                test_predictions,
                labels=[0, 1],
                target_names=[
                    "HPV Negative",
                    "HPV Positive"
                ],
                digits=4,
                zero_division=0
            )
        )

        print("Confusion matrix [[TN, FP], [FN, TP]]:")
        print(cm)

        print(
            f"Test ROC-AUC={test_auc:.4f} | "
            f"PR-AUC={test_ap:.4f} | "
            f"BA={test_ba:.4f} | "
            f"F1={test_f1:.4f}"
        )

        fold_results.append({
            "Fold": fold,
            "Checkpoint_Selected_By": name,
            "Best_Epoch": best_info[name]["epoch"],
            "Best_Validation_Metric": best_info[name]["value"],
            "Validation_Threshold": threshold,
            "Test_ROC_AUC": test_auc,
            "Test_PR_AUC": test_ap,
            "Test_Balanced_Accuracy": test_ba,
            "Test_F1": test_f1,
            "TN": cm[0, 0],
            "FP": cm[0, 1],
            "FN": cm[1, 0],
            "TP": cm[1, 1],
            "Test_N": len(test_idx)
        })

        # Store out-of-fold predictions.
        oof_scores[name][test_idx] = test_scores

        oof_predictions[name][test_idx] = test_predictions

        # Save the fold's classifier checkpoint.
        torch.save(
            best_states[name],
            OUT_DIR / f"fold_{fold}_{name}_classifier.pth"
        )

        # Save individual fold predictions.
        pd.DataFrame({
            "True_Label": test_labels,
            "HPV_Positive_Score": test_scores,
            "Predicted_Label": test_predictions,
            "Threshold": threshold
        }).to_csv(
            OUT_DIR / f"fold_{fold}_{name}_predictions.csv",
            index=False
        )

    # Save training history for this fold.
    pd.DataFrame(history).to_csv(
        OUT_DIR / f"fold_{fold}_training_history.csv",
        index=False
    )

    # Free GPU memory.
    del model
    del optimizer
    del train_loader
    del val_loader
    del test_loader

    del train_ds
    del val_ds
    del test_ds

    del best_states

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ============================================================
# 16. VERIFY THAT EVERY PATIENT WAS HELD OUT EXACTLY ONCE
# ============================================================

print("\n" + "=" * 75)
print("CROSS-VALIDATION COVERAGE CHECK")
print("=" * 75)

oof_fold = np.full(len(df), -1, dtype=int)

for fold, (_, test_idx) in enumerate(
    skf.split(X, y),
    start=1
):
    oof_fold[test_idx] = fold

assert (oof_fold >= 1).all()

for name in CHECKPOINT_NAMES:

    assert not np.isnan(oof_scores[name]).any()

    assert (oof_predictions[name] >= 0).all()

print("Total valid labelled patients:", len(df))
print("Patients assigned to held-out folds:", len(oof_fold))

print("\nPatients per fold:")
print(pd.Series(oof_fold).value_counts().sort_index())

print("\nEvery valid patient was held out exactly once.")


# ============================================================
# 17. INDIVIDUAL FOLD RESULTS
# ============================================================

fold_results_df = pd.DataFrame(fold_results)

fold_results_df.to_csv(
    OUT_DIR / "3fold_individual_results.csv",
    index=False
)

print("\nIndividual fold results:")
print(fold_results_df.to_string(index=False))


# ============================================================
# 18. MEAN AND SAMPLE SD ACROSS THREE FOLDS
# ============================================================

metric_columns = [
    "Test_ROC_AUC",
    "Test_PR_AUC",
    "Test_Balanced_Accuracy",
    "Test_F1"
]

summary_rows = []

for checkpoint in CHECKPOINT_NAMES:

    subset = fold_results_df[
        fold_results_df["Checkpoint_Selected_By"] == checkpoint
    ]

    for metric in metric_columns:

        mean_value = subset[metric].mean()

        # Sample standard deviation, ddof=1.
        sd_value = subset[metric].std(ddof=1)

        summary_rows.append({
            "Checkpoint_Selected_By": checkpoint,
            "Metric": metric,
            "Mean": mean_value,
            "SD": sd_value,
            "Mean +/- SD": (
                f"{mean_value:.4f} +/- {sd_value:.4f}"
            )
        })

summary_df = pd.DataFrame(summary_rows)

print("\n" + "=" * 75)
print("3-FOLD CROSS-VALIDATION: MEAN +/- SAMPLE SD")
print("=" * 75)

print(summary_df.to_string(index=False))

summary_df.to_csv(
    OUT_DIR / "3fold_mean_sd_results.csv",
    index=False
)


# ============================================================
# 19. POOLED OUT-OF-FOLD METRICS
# ============================================================

pooled_rows = []

for name in CHECKPOINT_NAMES:

    scores = oof_scores[name]
    predictions = oof_predictions[name]

    pooled_auc = roc_auc_score(y, scores)

    pooled_ap = average_precision_score(y, scores)

    pooled_ba = balanced_accuracy_score(
        y,
        predictions
    )

    pooled_f1 = f1_score(
        y,
        predictions,
        zero_division=0
    )

    pooled_cm = confusion_matrix(
        y,
        predictions,
        labels=[0, 1]
    )

    pooled_rows.append({
        "Checkpoint_Selected_By": name,
        "Pooled_ROC_AUC": pooled_auc,
        "Pooled_PR_AUC": pooled_ap,
        "Pooled_Balanced_Accuracy": pooled_ba,
        "Pooled_F1": pooled_f1,
        "TN": pooled_cm[0, 0],
        "FP": pooled_cm[0, 1],
        "FN": pooled_cm[1, 0],
        "TP": pooled_cm[1, 1],
        "Total_Patients": len(df)
    })

    print("\n" + "-" * 65)
    print("POOLED OUT-OF-FOLD RESULTS:", name)

    print(f"ROC-AUC: {pooled_auc:.4f}")
    print(f"PR-AUC: {pooled_ap:.4f}")
    print(f"Balanced Accuracy: {pooled_ba:.4f}")
    print(f"F1: {pooled_f1:.4f}")

    print("\nConfusion matrix:")
    print(pooled_cm)


pooled_df = pd.DataFrame(pooled_rows)

pooled_df.to_csv(
    OUT_DIR / "3fold_pooled_metrics.csv",
    index=False
)


# ============================================================
# 20. POOLED OUT-OF-FOLD ROC CURVES
# ============================================================

plt.figure(figsize=(8, 7))

for name in CHECKPOINT_NAMES:

    scores = oof_scores[name]

    fpr, tpr, _ = roc_curve(y, scores)

    auc = roc_auc_score(y, scores)

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{name} (AUC={auc:.4f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    "--",
    color="gray",
    label="Random classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "HECKTOR GPT-2 — 3-Fold Pooled ROC Curves"
)

plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    OUT_DIR / "3fold_pooled_ROC_all_checkpoints.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 21. POOLED OUT-OF-FOLD PRECISION-RECALL CURVES
# ============================================================

plt.figure(figsize=(8, 7))

for name in CHECKPOINT_NAMES:

    scores = oof_scores[name]

    precision, recall, _ = precision_recall_curve(
        y,
        scores
    )

    ap = average_precision_score(y, scores)

    plt.plot(
        recall,
        precision,
        linewidth=2,
        label=f"{name} (AP={ap:.4f})"
    )

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title(
    "HECKTOR GPT-2 — 3-Fold Pooled Precision-Recall Curves"
)

plt.legend(loc="best")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    OUT_DIR / "3fold_pooled_PR_all_checkpoints.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 22. SAVE ALL OUT-OF-FOLD PREDICTIONS
# ============================================================

oof_data = pd.DataFrame({
    "Patient_Index": np.arange(len(df)),
    "True_Label": y,
    "Held_Out_Fold": oof_fold
})

for name in CHECKPOINT_NAMES:

    safe_name = name.lower()

    oof_data[f"{safe_name}_probability"] = (
        oof_scores[name]
    )

    oof_data[f"{safe_name}_prediction"] = (
        oof_predictions[name]
    )

oof_data.to_csv(
    OUT_DIR / "3fold_all_out_of_fold_predictions.csv",
    index=False
)


# ============================================================
# 23. SAVE MODEL SETTINGS
# ============================================================

settings = {
    "Dataset": "HECKTOR",
    "Dataset_File": "HPV2025.xlsx",
    "Model": MODEL_NAME,
    "Features": ", ".join(FEATURES),
    "Number_of_Features": len(FEATURES),
    "Target": "HPV Status",
    "Number_of_Valid_Patients": len(df),
    "Number_of_Folds": N_FOLDS,
    "Validation_Size_Within_Outer_Training": VALIDATION_SIZE,
    "GPT2_Backbone_Frozen": True,
    "Trainable_Parameters": trainable_params,
    "Batch_Size": BATCH_SIZE,
    "Epochs": EPOCHS,
    "Learning_Rate": LEARNING_RATE,
    "Weight_Decay": WEIGHT_DECAY,
    "Dropout": DROPOUT,
    "Optimizer": "AdamW — classifier head only",
    "Class_Weighted_Loss": True,
    "Checkpoint_Selection": (
        "Validation Loss, ROC-AUC, PR-AUC, "
        "Balanced Accuracy"
    ),
    "Threshold_Selection": (
        "Maximum validation balanced accuracy; "
        "ties closest to 0.5"
    ),
    "Evaluation": "Stratified 3-fold cross-validation",
    "Mean_SD": "Mean and sample SD (ddof=1)"
}

pd.DataFrame([settings]).to_csv(
    OUT_DIR / "3fold_settings.csv",
    index=False
)


# ============================================================
# 24. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 75)
print("HECKTOR GPT-2 — 3-FOLD CROSS-VALIDATION COMPLETE")
print("=" * 75)

print("\nTotal patients used:", len(df))
print("Number of folds:", N_FOLDS)
print("GPT-2 backbone completely frozen: YES")
print("Only classifier head trained: YES")
print("Batch size:", BATCH_SIZE)
print("Epochs per fold:", EPOCHS)

print("\nResults saved to:")
print(OUT_DIR.resolve())

print("\nMain result files:")
print("1. 3fold_individual_results.csv")
print("2. 3fold_mean_sd_results.csv")
print("3. 3fold_pooled_metrics.csv")
print("4. 3fold_all_out_of_fold_predictions.csv")
print("5. 3fold_pooled_ROC_all_checkpoints.png")
print("6. 3fold_pooled_PR_all_checkpoints.png")
print("7. 3fold_settings.csv")

print("\nAll folds completed successfully.")
```
