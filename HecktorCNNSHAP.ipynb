```python
# ============================================================
# CNN — HECKTOR HPV CLASSIFIER
# Hancock-aligned methodology | Seed 42 | No SMOTE
# 8 predictors | 65/15/20 split | 500 epochs
# Four validation-selected checkpoints
# ============================================================

!pip -q install gdown

import os
import json
import random
import copy
import warnings
import joblib
import shutil

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gdown

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
MAX_EPOCHS = 500
BATCH_SIZE = 32


# ============================================================
# 1. REPRODUCIBILITY AND DEVICE
# ============================================================

def seed_everything(seed=42):
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)


# ============================================================
# 2. DOWNLOAD AND LOAD HECKTOR DATA
# ============================================================

# Download directly; no Google Drive mount is required.
BASE = "/content"
DATA_PATH = "/content/HPV2025.xlsx"
OUT_DIR = os.path.join(BASE, "CNN_results_HECKTOR")
CHECKPOINT_DIR = os.path.join(OUT_DIR, "checkpoints")

os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

FILE_ID = "1DGDH2e6dEkpdwwtCvH6ds1CwJT2Hdmkv"

if not os.path.isfile(DATA_PATH):
    print("Downloading HPV2025.xlsx from Google Drive...")

    downloaded = gdown.download(
        id=FILE_ID,
        output=DATA_PATH,
        quiet=False
    )

    if downloaded is None or not os.path.isfile(DATA_PATH):
        raise FileNotFoundError(
            "Could not download HPV2025.xlsx. "
            "Check the Google Drive file ID and sharing permissions."
        )

if not os.path.isfile(DATA_PATH) or os.path.getsize(DATA_PATH) == 0:
    raise FileNotFoundError(
        f"Dataset is missing or empty: {DATA_PATH}"
    )

print("Dataset ready:", DATA_PATH)
print("Dataset size (KB):", round(os.path.getsize(DATA_PATH) / 1024, 2))

df = pd.read_excel(DATA_PATH)

target_col = "HPV Status"

if target_col not in df.columns:
    raise KeyError(
        f"Missing target column '{target_col}'. "
        f"Available columns: {df.columns.tolist()}"
    )

# Remove rows without a target before modelling.
df = df.dropna(subset=[target_col]).copy()

# Preserve patient identifiers for result exports only.
id_col = next(
    (c for c in ["PatientID", "patient_id"] if c in df.columns),
    None
)

if id_col is None:
    patient_ids = pd.Series(
        np.arange(len(df)),
        index=df.index,
        name="patient_id"
    )
else:
    patient_ids = df[id_col].copy()

# Do not use IDs or unrelated task columns as predictors.
df = df.drop(
    columns=[
        "PatientID",
        "patient_id",
        "CenterID",
        "Task 1",
        "Task 2",
        "Task 3"
    ],
    errors="ignore"
)

# The eight specified HECKTOR predictors.
features = [
    "Age",
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

numeric_features = ["Age"]

categorical_features = [
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

missing_features = [
    col for col in features if col not in df.columns
]

if missing_features:
    raise KeyError(
        f"Missing predictor columns: {missing_features}\n"
        f"Available columns: {df.columns.tolist()}"
    )


# ------------------------------------------------------------
# Normalize HPV Status to binary labels: negative=0, positive=1
# ------------------------------------------------------------

def normalize_label(value):
    if pd.isna(value):
        return np.nan

    value_str = str(value).strip().lower()

    if value_str in {
        "negative", "neg", "0", "0.0", "false", "no"
    }:
        return 0

    if value_str in {
        "positive", "pos", "1", "1.0", "true", "yes"
    }:
        return 1

    try:
        number = float(value_str)
        if number in (0.0, 1.0):
            return int(number)
    except (ValueError, TypeError):
        pass

    return np.nan


df["HPV_Label"] = df[target_col].map(normalize_label)
df = df.dropna(subset=["HPV_Label"]).copy()
df["HPV_Label"] = df["HPV_Label"].astype(int)

if df["HPV_Label"].nunique() != 2:
    raise ValueError(
        "HPV Status must contain both classes, 0 and 1, "
        "after label normalization."
    )

# Identifiers are retained for output but never enter the model.
df = df[features + ["HPV_Label"]].copy()
df["patient_id"] = patient_ids.reindex(df.index).values

print("Rows available:", len(df))
print("Predictors:", features)
print("SMOTE: False")
print("Target class counts:")
print(df["HPV_Label"].value_counts().sort_index())


# ============================================================
# 3. STRATIFIED TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# Same proportions and random seed as Hancock:
# 80% development data / 20% test,
# then 15% of the full dataset for validation.
train_val, test = train_test_split(
    df,
    test_size=0.20,
    random_state=SEED,
    stratify=df["HPV_Label"]
)

train, val = train_test_split(
    train_val,
    test_size=0.1875,
    random_state=SEED,
    stratify=train_val["HPV_Label"]
)

train = train.reset_index(drop=True)
val = val.reset_index(drop=True)
test = test.reset_index(drop=True)

for name, frame in [
    ("Train", train),
    ("Validation", val),
    ("Test", test)
]:
    print(
        f"{name}: {len(frame)} rows | "
        f"class counts: "
        f"{frame['HPV_Label'].value_counts().sort_index().to_dict()}"
    )


# ============================================================
# 4. IMPUTATION, ONE-HOT ENCODING AND SCALING
# ============================================================

X_train_df = train[features].copy()
X_val_df = val[features].copy()
X_test_df = test[features].copy()

# Clean numeric and categorical values.
for frame in [X_train_df, X_val_df, X_test_df]:

    for col in numeric_features:
        frame[col] = pd.to_numeric(
            frame[col],
            errors="coerce"
        )

    for col in categorical_features:
        frame[col] = (
            frame[col]
            .astype("object")
            .replace(r"^\s*$", np.nan, regex=True)
            .replace(["nan", "None", "NaN"], np.nan)
        )

# Fit imputation values using training data only.
for col in numeric_features:
    median_value = X_train_df[col].median()

    if pd.isna(median_value):
        median_value = 0.0

    for frame in [X_train_df, X_val_df, X_test_df]:
        frame[col] = frame[col].fillna(median_value)

for col in categorical_features:
    mode_values = X_train_df[col].mode(dropna=True)

    fill_value = (
        str(mode_values.iloc[0])
        if not mode_values.empty
        else "Unknown"
    )

    for frame in [X_train_df, X_val_df, X_test_df]:
        frame[col] = (
            frame[col]
            .fillna(fill_value)
            .astype(str)
            .replace(["", "nan", "None", "NaN"], "Unknown")
        )

# One-hot encode categorical predictors.
try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    # Compatibility with older scikit-learn versions.
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

# Standardize Age; one-hot encode all other predictors.
preprocessor = ColumnTransformer([
    (
        "numeric",
        StandardScaler(),
        numeric_features
    ),
    (
        "categorical",
        encoder,
        categorical_features
    )
])

# Fit on training data only; transform validation and test.
X_train = preprocessor.fit_transform(
    X_train_df
).astype(np.float32)

X_val = preprocessor.transform(
    X_val_df
).astype(np.float32)

X_test = preprocessor.transform(
    X_test_df
).astype(np.float32)

y_train = train["HPV_Label"].to_numpy(dtype=np.int64)
y_val = val["HPV_Label"].to_numpy(dtype=np.int64)
y_test = test["HPV_Label"].to_numpy(dtype=np.int64)

print("Original predictor count:", len(features))
print("Encoded feature count:", X_train.shape[1])


# ============================================================
# 5. AUTOMATIC CLASS WEIGHTS — NO SMOTE
# ============================================================

counts = np.bincount(y_train, minlength=2)

if np.any(counts == 0):
    raise ValueError(
        "Both HPV classes must be present in the training split."
    )

# Inverse-frequency class weights calculated on training data only.
weights = len(y_train) / (2.0 * counts)

class_weights = torch.tensor(
    weights,
    dtype=torch.float32,
    device=device
)

print("Automatic class weights:", weights)
print("SMOTE: False")


# ============================================================
# 6. TENSORS AND TRAINING DATA LOADER
# ============================================================

train_dataset = TensorDataset(
    torch.tensor(X_train, dtype=torch.float32),
    torch.tensor(y_train, dtype=torch.long)
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    generator=torch.Generator().manual_seed(SEED)
)

X_val_t = torch.tensor(
    X_val,
    dtype=torch.float32,
    device=device
)

y_val_t = torch.tensor(
    y_val,
    dtype=torch.long,
    device=device
)

X_test_t = torch.tensor(
    X_test,
    dtype=torch.float32,
    device=device
)


# ============================================================
# 7. DEFINE HANCOCK-ALIGNED CNN ARCHITECTURE
# ============================================================

class CNNClassifier(nn.Module):

    def __init__(self, input_dim):
        super().__init__()

        self.conv = nn.Conv1d(
            in_channels=1,
            out_channels=16,
            kernel_size=3,
            padding=1
        )

        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.25)

        self.fc1 = nn.Linear(input_dim * 16, 32)
        self.fc2 = nn.Linear(32, 16)
        self.fc3 = nn.Linear(16, 2)

    def forward(self, x):
        # Shape: (batch, predictors)
        x = x.unsqueeze(1)

        x = self.conv(x)
        x = self.relu(x)
        x = self.dropout(x)

        x = x.flatten(start_dim=1)

        x = self.fc1(x)
        x = self.relu(x)

        x = self.fc2(x)
        x = self.relu(x)

        x = self.fc3(x)

        return x


model = CNNClassifier(
    input_dim=X_train.shape[1]
).to(device)

criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    reduction="none"
)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0001,
    weight_decay=1e-4
)

print(model)


# ============================================================
# 8. VALIDATION HELPERS AND CHECKPOINT RULES
# ============================================================

thresholds = np.linspace(0.10, 0.90, 161)


def select_threshold(y_true, probabilities):
    """Maximize validation balanced accuracy; ties closest to 0.5."""

    scores = [
        balanced_accuracy_score(
            y_true,
            (probabilities >= threshold).astype(int)
        )
        for threshold in thresholds
    ]

    best_score = max(scores)

    best_threshold = min(
        (
            threshold
            for threshold, score in zip(thresholds, scores)
            if np.isclose(score, best_score)
        ),
        key=lambda threshold: abs(threshold - 0.5)
    )

    return float(best_threshold), float(best_score)


def predict_probabilities(model, X_tensor):
    model.eval()

    with torch.no_grad():
        return torch.softmax(
            model(X_tensor),
            dim=1
        )[:, 1].cpu().numpy()


def validation_metrics(model):
    probabilities = predict_probabilities(model, X_val_t)

    threshold, balanced_acc = select_threshold(
        y_val,
        probabilities
    )

    roc_auc = (
        roc_auc_score(y_val, probabilities)
        if len(np.unique(y_val)) == 2
        else np.nan
    )

    pr_auc = average_precision_score(
        y_val,
        probabilities
    )

    return {
        "probabilities": probabilities,
        "threshold": threshold,
        "balanced_accuracy": float(balanced_acc),
        "roc_auc": float(roc_auc),
        "pr_auc": float(pr_auc)
    }


checkpoint_specs = {
    "best_val_loss": {
        "filename": "cnn_best_val_loss.pt"
    },
    "best_val_roc_auc": {
        "filename": "cnn_best_val_roc_auc.pt"
    },
    "best_val_pr_auc": {
        "filename": "cnn_best_val_pr_auc.pt"
    },
    "best_val_balanced_accuracy": {
        "filename": "cnn_best_val_balanced_accuracy.pt"
    }
}

best_values = {
    "best_val_loss": float("inf"),
    "best_val_roc_auc": -float("inf"),
    "best_val_pr_auc": -float("inf"),
    "best_val_balanced_accuracy": -float("inf")
}

best_epochs = {}
history = []


def save_checkpoint(name, epoch, val_loss, result):
    checkpoint = {
        "model_state_dict": copy.deepcopy(model.state_dict()),
        "epoch": int(epoch),
        "validation_loss": float(val_loss),
        "validation_roc_auc": float(result["roc_auc"]),
        "validation_pr_auc": float(result["pr_auc"]),
        "validation_balanced_accuracy": float(
            result["balanced_accuracy"]
        ),
        "validation_threshold": float(result["threshold"])
    }

    torch.save(
        checkpoint,
        os.path.join(
            CHECKPOINT_DIR,
            checkpoint_specs[name]["filename"]
        )
    )


# ============================================================
# 9. TRAIN FOR ALL 500 EPOCHS — NO EARLY STOPPING
# ============================================================

for epoch in range(1, MAX_EPOCHS + 1):

    model.train()
    running_loss = 0.0
    seen = 0

    for xb, yb in train_loader:
        xb = xb.to(device)
        yb = yb.to(device)

        optimizer.zero_grad()

        logits = model(xb)

        per_example_loss = criterion(logits, yb)
        loss = per_example_loss.mean()

        loss.backward()
        optimizer.step()

        running_loss += per_example_loss.detach().sum().item()
        seen += len(yb)

    train_loss = running_loss / max(seen, 1)

    # Validation after each epoch.
    model.eval()

    with torch.no_grad():
        val_loss = criterion(
            model(X_val_t),
            y_val_t
        ).mean().item()

    result = validation_metrics(model)

    history.append({
        "epoch": epoch,
        "train_loss": float(train_loss),
        "val_loss": float(val_loss),
        "val_roc_auc": result["roc_auc"],
        "val_pr_auc": result["pr_auc"],
        "val_balanced_accuracy": result["balanced_accuracy"],
        "val_threshold": result["threshold"]
    })

    # Checkpoint 1: lowest validation loss.
    if val_loss < best_values["best_val_loss"]:
        best_values["best_val_loss"] = float(val_loss)
        best_epochs["best_val_loss"] = epoch

        save_checkpoint(
            "best_val_loss",
            epoch,
            val_loss,
            result
        )

    # Checkpoint 2: highest validation ROC-AUC.
    if (
        np.isfinite(result["roc_auc"])
        and result["roc_auc"] > best_values["best_val_roc_auc"]
    ):
        best_values["best_val_roc_auc"] = float(result["roc_auc"])
        best_epochs["best_val_roc_auc"] = epoch

        save_checkpoint(
            "best_val_roc_auc",
            epoch,
            val_loss,
            result
        )

    # Checkpoint 3: highest validation PR-AUC.
    if (
        np.isfinite(result["pr_auc"])
        and result["pr_auc"] > best_values["best_val_pr_auc"]
    ):
        best_values["best_val_pr_auc"] = float(result["pr_auc"])
        best_epochs["best_val_pr_auc"] = epoch

        save_checkpoint(
            "best_val_pr_auc",
            epoch,
            val_loss,
            result
        )

    # Checkpoint 4: highest validation balanced accuracy.
    if (
        result["balanced_accuracy"]
        > best_values["best_val_balanced_accuracy"]
    ):
        best_values["best_val_balanced_accuracy"] = float(
            result["balanced_accuracy"]
        )
        best_epochs["best_val_balanced_accuracy"] = epoch

        save_checkpoint(
            "best_val_balanced_accuracy",
            epoch,
            val_loss,
            result
        )

    if epoch == 1 or epoch % 25 == 0:
        print(
            f"Epoch {epoch:03d}/{MAX_EPOCHS} | "
            f"train loss={train_loss:.4f} | "
            f"val loss={val_loss:.4f} | "
            f"val ROC-AUC={result['roc_auc']:.4f} | "
            f"val PR-AUC={result['pr_auc']:.4f} | "
            f"val balanced accuracy="
            f"{result['balanced_accuracy']:.4f}"
        )

if len(history) != MAX_EPOCHS:
    raise RuntimeError(
        f"Expected {MAX_EPOCHS} epochs, but ran {len(history)}."
    )

print("Training completed:", len(history), "epochs")
print("Best checkpoint epochs:", best_epochs)

history_df = pd.DataFrame(history)

history_df.to_csv(
    os.path.join(OUT_DIR, "training_history.csv"),
    index=False
)

history_df.to_csv(
    os.path.join(OUT_DIR, "CNN_epoch_history.csv"),
    index=False
)


# ============================================================
# 10. PLOT TRAINING HISTORY
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    history_df["epoch"],
    history_df["train_loss"],
    label="Train loss"
)

plt.plot(
    history_df["epoch"],
    history_df["val_loss"],
    label="Validation loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("CNN — HECKTOR Training History")
plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(OUT_DIR, "training_history.png"),
    dpi=150
)

plt.show()


# ============================================================
# 11. EVALUATE ALL FOUR CHECKPOINTS ON THE TEST SET
# ============================================================

all_test_metrics = []
all_test_predictions = []
checkpoint_summary = []

for checkpoint_name, spec in checkpoint_specs.items():

    checkpoint_path = os.path.join(
        CHECKPOINT_DIR,
        spec["filename"]
    )

    try:
        checkpoint = torch.load(
            checkpoint_path,
            map_location=device,
            weights_only=False
        )
    except TypeError:
        checkpoint = torch.load(
            checkpoint_path,
            map_location=device
        )

    model.load_state_dict(checkpoint["model_state_dict"])

    # Threshold is selected using validation data only.
    val_prob = predict_probabilities(model, X_val_t)

    threshold, val_ba = select_threshold(
        y_val,
        val_prob
    )

    # Evaluate the selected checkpoint on the test set.
    test_prob = predict_probabilities(model, X_test_t)
    test_pred = (test_prob >= threshold).astype(int)

    test_auc = (
        roc_auc_score(y_test, test_prob)
        if len(np.unique(y_test)) == 2
        else np.nan
    )

    test_ap = average_precision_score(
        y_test,
        test_prob
    )

    metrics = {
        "Checkpoint": checkpoint_name,
        "Checkpoint_Epoch": int(checkpoint["epoch"]),
        "Validation_Loss": float(checkpoint["validation_loss"]),
        "Validation_ROC_AUC": float(checkpoint["validation_roc_auc"]),
        "Validation_PR_AUC": float(checkpoint["validation_pr_auc"]),
        "Validation_Balanced_Accuracy": float(val_ba),
        "Threshold": float(threshold),
        "Accuracy": accuracy_score(y_test, test_pred),
        "Balanced_Accuracy": balanced_accuracy_score(y_test, test_pred),
        "Precision": precision_score(
            y_test, test_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, test_pred, zero_division=0
        ),
        "F1": f1_score(
            y_test, test_pred, zero_division=0
        ),
        "ROC_AUC": test_auc,
        "PR_AUC": test_ap
    }

    all_test_metrics.append(metrics)

    all_test_predictions.append(
        pd.DataFrame({
            "patient_id": test["patient_id"].to_numpy(),
            "y_true": y_test,
            "probability_positive": test_prob,
            "prediction": test_pred,
            "checkpoint": checkpoint_name,
            "checkpoint_epoch": int(checkpoint["epoch"]),
            "threshold": float(threshold)
        })
    )

    checkpoint_summary.append({
        "Checkpoint": checkpoint_name,
        "Epoch": int(checkpoint["epoch"]),
        "Validation_Loss": float(checkpoint["validation_loss"]),
        "Validation_ROC_AUC": float(checkpoint["validation_roc_auc"]),
        "Validation_PR_AUC": float(checkpoint["validation_pr_auc"]),
        "Validation_Balanced_Accuracy": float(val_ba),
        "Threshold": float(threshold)
    })

    print("\n" + "=" * 65)
    print("CHECKPOINT:", checkpoint_name)
    print("Epoch:", checkpoint["epoch"])
    print("Threshold:", round(threshold, 3))
    print("\nTEST METRICS")
    print(pd.Series(metrics).to_string())
    print("\nConfusion matrix:")
    print(confusion_matrix(y_test, test_pred))
    print("\nClassification report:")
    print(
        classification_report(
            y_test,
            test_pred,
            zero_division=0
        )
    )


metrics_df = pd.DataFrame(all_test_metrics)

metrics_df.to_csv(
    os.path.join(OUT_DIR, "metrics_all_checkpoints.csv"),
    index=False
)

metrics_df.to_csv(
    os.path.join(OUT_DIR, "CNN_results.csv"),
    index=False
)

metrics_df.to_csv(
    os.path.join(OUT_DIR, "metrics.csv"),
    index=False
)

predictions_df = pd.concat(
    all_test_predictions,
    ignore_index=True
)

predictions_df.to_csv(
    os.path.join(OUT_DIR, "test_predictions_all_checkpoints.csv"),
    index=False
)

predictions_df.to_csv(
    os.path.join(OUT_DIR, "CNN_test_predictions.csv"),
    index=False
)

predictions_df.to_csv(
    os.path.join(OUT_DIR, "test_predictions.csv"),
    index=False
)

checkpoint_summary_df = pd.DataFrame(checkpoint_summary)

checkpoint_summary_df.to_csv(
    os.path.join(OUT_DIR, "checkpoint_summary.csv"),
    index=False
)


# ============================================================
# 12. TEST ROC CURVES FOR ALL FOUR CHECKPOINTS
# ============================================================

plt.figure(figsize=(8, 6))

for name, spec in checkpoint_specs.items():

    checkpoint_path = os.path.join(
        CHECKPOINT_DIR,
        spec["filename"]
    )

    try:
        checkpoint = torch.load(
            checkpoint_path,
            map_location=device,
            weights_only=False
        )
    except TypeError:
        checkpoint = torch.load(
            checkpoint_path,
            map_location=device
        )

    model.load_state_dict(checkpoint["model_state_dict"])
    probabilities = predict_probabilities(model, X_test_t)

    if len(np.unique(y_test)) == 2:
        fpr, tpr, _ = roc_curve(y_test, probabilities)
        auc_value = roc_auc_score(y_test, probabilities)

        plt.plot(
            fpr,
            tpr,
            label=f"{name} (AUC={auc_value:.3f})"
        )

plt.plot([0, 1], [0, 1], "--", label="Chance")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("CNN — HECKTOR Test ROC Curves")
plt.legend(fontsize=8)
plt.tight_layout()

plt.savefig(
    os.path.join(OUT_DIR, "roc_curves_all_checkpoints.png"),
    dpi=150
)

plt.savefig(
    os.path.join(OUT_DIR, "roc_curve.png"),
    dpi=150
)

plt.show()


# ============================================================
# 13. TEST PRECISION-RECALL CURVES FOR ALL CHECKPOINTS
# ============================================================

plt.figure(figsize=(8, 6))

for name, spec in checkpoint_specs.items():

    checkpoint_path = os.path.join(
        CHECKPOINT_DIR,
        spec["filename"]
    )

    try:
        checkpoint = torch.load(
            checkpoint_path,
            map_location=device,
            weights_only=False
        )
    except TypeError:
        checkpoint = torch.load(
            checkpoint_path,
            map_location=device
        )

    model.load_state_dict(checkpoint["model_state_dict"])
    probabilities = predict_probabilities(model, X_test_t)

    precision_values, recall_values, _ = precision_recall_curve(
        y_test,
        probabilities
    )

    ap_value = average_precision_score(
        y_test,
        probabilities
    )

    plt.plot(
        recall_values,
        precision_values,
        label=f"{name} (AP={ap_value:.3f})"
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("CNN — HECKTOR Test Precision-Recall Curves")
plt.legend(fontsize=8)
plt.tight_layout()

plt.savefig(
    os.path.join(OUT_DIR, "pr_curves_all_checkpoints.png"),
    dpi=150
)

plt.savefig(
    os.path.join(OUT_DIR, "pr_curve.png"),
    dpi=150
)

plt.show()


# ============================================================
# 14. SAVE PREPROCESSOR AND SETTINGS
# ============================================================

joblib.dump(
    preprocessor,
    os.path.join(OUT_DIR, "cnn_preprocessor.joblib")
)

settings = {
    "model": "CNN",
    "dataset": "HECKTOR",
    "seed": SEED,
    "features": features,
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "removed_features": ["Relapse", "RFS", "Treatment"],
    "target_column": target_col,
    "target_mapping": {
        "Negative": 0,
        "Positive": 1
    },
    "class_weights": weights.tolist(),
    "smote": False,
    "early_stopping": False,
    "max_epochs": MAX_EPOCHS,
    "epochs_run": len(history),
    "batch_size": BATCH_SIZE,
    "architecture": (
        "Conv1d(1,16,kernel_size=3,padding=1)"
        "-Dropout(0.25)-FC32-FC16-FC2"
    ),
    "optimizer": "Adam",
    "learning_rate": 0.0001,
    "weight_decay": 0.0001,
    "train_size": len(train),
    "validation_size": len(val),
    "test_size": len(test),
    "checkpoint_epochs": {
        key: int(value)
        for key, value in best_epochs.items()
    },
    "checkpoint_files": {
        key: value["filename"]
        for key, value in checkpoint_specs.items()
    },
    "threshold_selection": (
        "Maximize validation balanced accuracy; "
        "ties resolved by threshold closest to 0.5"
    ),
    "test_used_for_model_selection": False
}

with open(
    os.path.join(OUT_DIR, "settings.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(settings, f, indent=2)


# ============================================================
# 15. FINAL CHECKPOINT COMPARISON AND PACKAGE RESULTS
# ============================================================

display_columns = [
    "Checkpoint",
    "Checkpoint_Epoch",
    "Threshold",
    "Accuracy",
    "Balanced_Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC",
    "PR_AUC"
]

print("\n" + "=" * 80)
print("CNN — HECKTOR TEST SET CHECKPOINT COMPARISON")
print("=" * 80)

print(metrics_df[display_columns].to_string(index=False))

print("\nSaved results to:", OUT_DIR)
print("Saved checkpoints to:", CHECKPOINT_DIR)

print("\nOutput files:")
for filename in sorted(os.listdir(OUT_DIR)):
    print(" -", filename)

print("\nCheckpoint files:")
for filename in sorted(os.listdir(CHECKPOINT_DIR)):
    print(" -", filename)

# Package results so they can be downloaded before the runtime resets.
RESULTS_ZIP = shutil.make_archive(
    OUT_DIR,
    "zip",
    root_dir=OUT_DIR
)

print("\nResults folder:", OUT_DIR)
print("Results archive:", RESULTS_ZIP)
print(
    "IMPORTANT: /content is temporary in Colab. "
    "Download the ZIP before resetting the runtime."
)

# In a separate Colab cell, run this to download the ZIP:
# from google.colab import files
# files.download(RESULTS_ZIP)
```

```python
# ------------------------------------------------------------
# 4. SHAP BEESWARM PLOT — CORRECTED CNN INPUT DIMENSIONS
# ------------------------------------------------------------

print("\nGenerating SHAP explanations...")

# Background and test samples remain 2D:
# (samples, encoded features)
background_size = min(50, len(X_train_encoded))
explain_size = min(100, len(X_test_encoded))

rng = np.random.default_rng(SEED)

background = X_train_encoded[
    rng.choice(
        len(X_train_encoded),
        background_size,
        replace=False
    )
]

X_explain = X_test_encoded[:explain_size]


# CNN expects: batch x 1 x encoded_features
# SHAP receives 2D input and the wrapper adds the channel dimension.
class CNNProbability(torch.nn.Module):

    def __init__(self, original_model):
        super().__init__()
        self.original_model = original_model

    def forward(self, x):

        # x arrives as (batch, encoded_features)
        if x.ndim == 2:
            x = x.unsqueeze(1)

        logits = self.original_model(x)

        return torch.softmax(logits, dim=1)[:, 1]


shap_model = CNNProbability(model).to(device)
shap_model.eval()

background_tensor = torch.tensor(
    background,
    dtype=torch.float32,
    device=device
)

explain_tensor = torch.tensor(
    X_explain,
    dtype=torch.float32,
    device=device
)

print("SHAP background shape:", background_tensor.shape)
print("SHAP explanation shape:", explain_tensor.shape)

# GradientExplainer
explainer = shap.GradientExplainer(
    shap_model,
    background_tensor
)

shap_values = explainer.shap_values(
    explain_tensor
)

# Convert SHAP output to NumPy
if isinstance(shap_values, list):
    shap_values = shap_values[0]

shap_values = np.asarray(shap_values)

# Remove singleton output dimensions if present
shap_values = np.squeeze(shap_values)

if shap_values.ndim == 1:
    shap_values = shap_values.reshape(1, -1)

if shap_values.shape[0] != len(X_explain):
    shap_values = shap_values.reshape(
        len(X_explain), -1
    )

print("SHAP values shape:", shap_values.shape)

# Get encoded feature names
encoded_feature_names = list(
    preprocessor.get_feature_names_out()
)

# Group encoded SHAP values into the 8 original predictors
grouped_shap = np.zeros(
    (len(X_explain), len(feature_names))
)

for j, original_feature in enumerate(feature_names):

    matching_columns = [
        i for i, encoded_name in enumerate(encoded_feature_names)
        if encoded_name.split("__")[-1] == original_feature
        or encoded_name.split("__")[-1].startswith(
            original_feature + "_"
        )
    ]

    if not matching_columns:
        matching_columns = [
            i for i, encoded_name in enumerate(encoded_feature_names)
            if original_feature in encoded_name
        ]

    if matching_columns:
        grouped_shap[:, j] = shap_values[
            :, matching_columns
        ].sum(axis=1)

# Generate beeswarm plot
plt.figure(figsize=(9, 6))

shap.summary_plot(
    grouped_shap,
    X_test_original.iloc[:explain_size].to_numpy(),
    feature_names=feature_names,
    plot_type="dot",
    show=False
)

plt.title("HECKTOR CNN – SHAP Beeswarm (Epoch 50)")
plt.tight_layout()

shap_png = os.path.join(
    OUTPUT_DIR,
    "HECKTOR_CNN_Epoch50_SHAP_Beeswarm.png"
)

plt.savefig(
    shap_png,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# Save SHAP values
pd.DataFrame(
    grouped_shap,
    columns=feature_names
).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_SHAP_Values.csv"
    ),
    index=False
)

print("SHAP beeswarm successfully saved:", shap_png)
```

```python
# ============================================================
# HECKTOR CNN — EPOCH 50
# SHAP BEESWARM + LIME GREEN/RED
# ROC + PR POINTS
# SAVE AND DOWNLOAD ALL RESULTS
# ============================================================

%pip -q install shap lime

import os
import zipfile
import numpy as np
import pandas as pd
import torch
import shap
import matplotlib.pyplot as plt

from lime.lime_tabular import LimeTabularExplainer
from sklearn.metrics import (
    roc_curve,
    precision_recall_curve,
    roc_auc_score,
    average_precision_score
)
from google.colab import files

# ------------------------------------------------------------
# 1. SETTINGS
# ------------------------------------------------------------

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

CHECKPOINT_PATH = (
    "/content/CNN_results_HECKTOR/checkpoints/"
    "cnn_best_val_roc_auc.pt"
)

OUTPUT_DIR = "/content/HECKTOR_CNN_EPOCH50_FINAL"
os.makedirs(OUTPUT_DIR, exist_ok=True)

feature_names = [
    "Age",
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

num_cols = [
    "Age",
    "T-stage",
    "N-stage",
    "M-stage"
]

cat_cols = [
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status"
]

# ------------------------------------------------------------
# 2. LOAD EPOCH 50 CHECKPOINT
# ------------------------------------------------------------

try:
    checkpoint = torch.load(
        CHECKPOINT_PATH,
        map_location=device,
        weights_only=False
    )
except TypeError:
    checkpoint = torch.load(
        CHECKPOINT_PATH,
        map_location=device
    )

if isinstance(checkpoint, dict):
    if "model_state_dict" in checkpoint:
        state_dict = checkpoint["model_state_dict"]
    elif "state_dict" in checkpoint:
        state_dict = checkpoint["state_dict"]
    else:
        state_dict = checkpoint
else:
    state_dict = checkpoint

model.load_state_dict(state_dict)
model = model.to(device)
model.eval()

print("Loaded:", CHECKPOINT_PATH)
print("Selected checkpoint: best_val_roc_auc")
print("Checkpoint epoch: 50")

# ------------------------------------------------------------
# 3. PREPARE ORIGINAL AND ENCODED DATA
# ------------------------------------------------------------

X_train_original = X_train_df[feature_names].copy()
X_test_original = X_test_df[feature_names].copy()

X_train_encoded = preprocessor.transform(X_train_original)
X_test_encoded = preprocessor.transform(X_test_original)

if hasattr(X_train_encoded, "toarray"):
    X_train_encoded = X_train_encoded.toarray()

if hasattr(X_test_encoded, "toarray"):
    X_test_encoded = X_test_encoded.toarray()

X_train_encoded = np.asarray(
    X_train_encoded, dtype=np.float32
)

X_test_encoded = np.asarray(
    X_test_encoded, dtype=np.float32
)

print("Training data:", X_train_encoded.shape)
print("Test data:", X_test_encoded.shape)

# ------------------------------------------------------------
# 4. CNN PREDICTION FUNCTION
# ------------------------------------------------------------

def predict_proba_encoded(X):

    X = np.asarray(X, dtype=np.float32)

    X_tensor = torch.tensor(
        X,
        dtype=torch.float32,
        device=device
    ).unsqueeze(1)

    model.eval()

    with torch.no_grad():
        logits = model(X_tensor)
        probabilities = torch.softmax(logits, dim=1)

    return probabilities.cpu().numpy()


# Check that predictions work before SHAP or LIME
test_probabilities = predict_proba_encoded(
    X_test_encoded[:5]
)

assert test_probabilities.shape == (5, 2)

print("Prediction function verified.")
print("Example predictions:", test_probabilities[:3])

# ------------------------------------------------------------
# 5. SHAP BEESWARM — KERNEL EXPLAINER
# ------------------------------------------------------------

print("\nGenerating SHAP explanations...")

# Use a small representative background sample.
# KernelExplainer works with the existing 2D prediction function.
rng = np.random.default_rng(SEED)

background_size = min(30, len(X_train_encoded))
explain_size = min(50, len(X_test_encoded))

background_indices = rng.choice(
    len(X_train_encoded),
    background_size,
    replace=False
)

background = X_train_encoded[background_indices]
X_explain = X_test_encoded[:explain_size]

# Explain the probability of Class 1
def predict_class1(X):
    return predict_proba_encoded(X)[:, 1]


kernel_explainer = shap.KernelExplainer(
    predict_class1,
    background
)

# nsamples controls SHAP computation time.
# Increase to 200 or 300 if you want a more detailed approximation.
raw_shap_values = kernel_explainer.shap_values(
    X_explain,
    nsamples=100,
    silent=False
)

shap_values = np.asarray(raw_shap_values)

# Handle different SHAP output shapes
if shap_values.ndim == 3:
    shap_values = shap_values[:, :, 0]

shap_values = shap_values.reshape(
    len(X_explain),
    X_train_encoded.shape[1]
)

print("Encoded SHAP shape:", shap_values.shape)

# ------------------------------------------------------------
# 6. GROUP ENCODED SHAP VALUES INTO 8 ORIGINAL FEATURES
# ------------------------------------------------------------

encoded_feature_names = list(
    preprocessor.get_feature_names_out()
)

grouped_shap = np.zeros(
    (len(X_explain), len(feature_names))
)

grouped_columns = {}

for j, original_feature in enumerate(feature_names):

    matching_columns = []

    for i, encoded_name in enumerate(encoded_feature_names):

        clean_name = encoded_name.split("__")[-1]

        if (
            clean_name == original_feature
            or clean_name.startswith(original_feature + "_")
        ):
            matching_columns.append(i)

    if not matching_columns:
        raise ValueError(
            f"Could not identify encoded columns for {original_feature}. "
            f"Check preprocessor.get_feature_names_out()."
        )

    grouped_columns[original_feature] = matching_columns

    # Sum one-hot contributions to original feature
    grouped_shap[:, j] = shap_values[
        :, matching_columns
    ].sum(axis=1)

print("\nSHAP feature grouping:")
for feature, columns in grouped_columns.items():
    print(feature, ":", len(columns), "encoded column(s)")

# Convert categorical display values to numeric codes
# so the beeswarm plot can colour observations.
X_display = X_test_original.iloc[:explain_size].copy()

for col in cat_cols:
    X_display[col] = pd.Categorical(
        X_display[col],
        categories=pd.Categorical(
            X_train_original[col]
        ).categories
    ).codes

X_display = X_display.astype(float)

# SHAP beeswarm plot
plt.figure(figsize=(10, 7))

shap.summary_plot(
    grouped_shap,
    X_display.to_numpy(),
    feature_names=feature_names,
    plot_type="dot",
    show=False
)

plt.title(
    "HECKTOR CNN – SHAP Beeswarm (Epoch 50)",
    fontsize=13
)

plt.tight_layout()

shap_png = os.path.join(
    OUTPUT_DIR,
    "HECKTOR_CNN_Epoch50_SHAP_Beeswarm.png"
)

plt.savefig(
    shap_png,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# Save grouped SHAP values
pd.DataFrame(
    grouped_shap,
    columns=feature_names
).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_SHAP_Values.csv"
    ),
    index=False
)

print("SHAP beeswarm saved:", shap_png)

# ------------------------------------------------------------
# 7. LIME GREEN AND RED PLOT
# ------------------------------------------------------------

print("\nGenerating LIME explanation...")

X_train_lime = X_train_original.copy()
X_test_lime = X_test_original.copy()

categorical_indices = []
categorical_mappings = {}
categorical_names = {}

for col in cat_cols:

    idx = feature_names.index(col)
    categorical_indices.append(idx)

    categories = list(
        X_train_original[col].dropna().unique()
    )

    mapping = {
        value: i for i, value in enumerate(categories)
    }

    categorical_mappings[col] = mapping

    categorical_names[idx] = {
        i: str(value)
        for value, i in mapping.items()
    }

    X_train_lime[col] = (
        X_train_original[col]
        .map(mapping)
        .fillna(0)
    )

    X_test_lime[col] = (
        X_test_original[col]
        .map(mapping)
        .fillna(0)
    )

X_train_lime = X_train_lime.astype(float)
X_test_lime = X_test_lime.astype(float)


# Reconstruct original categorical values from LIME codes
def lime_predict_proba(data):

    df = pd.DataFrame(
        data,
        columns=feature_names
    )

    for col in cat_cols:

        reverse_mapping = {
            code: value
            for value, code in categorical_mappings[col].items()
        }

        fallback_value = X_train_original[col].mode().iloc[0]

        df[col] = [
            reverse_mapping.get(
                int(round(value)),
                fallback_value
            )
            for value in df[col]
        ]

    encoded = preprocessor.transform(df)

    if hasattr(encoded, "toarray"):
        encoded = encoded.toarray()

    return predict_proba_encoded(
        np.asarray(encoded, dtype=np.float32)
    )


lime_explainer = LimeTabularExplainer(
    training_data=X_train_lime.to_numpy(),
    feature_names=feature_names,
    categorical_features=categorical_indices,
    categorical_names=categorical_names,
    mode="classification",
    class_names=["HPV Negative", "HPV Positive"],
    discretize_continuous=True,
    random_state=SEED
)

# Explain first test patient
patient_index = 0

lime_explanation = lime_explainer.explain_instance(
    data_row=X_test_lime.iloc[patient_index].to_numpy(),
    predict_fn=lime_predict_proba,
    num_features=8,
    num_samples=3000
)

lime_contributions = lime_explanation.as_list(
    label=1
)

assert len(lime_contributions) > 0

print("\nLIME contributions for test patient", patient_index)

for feature, weight in lime_contributions:
    print(f"{feature}: {weight:.6f}")

# Green = positive contribution
# Red = negative contribution
lime_labels = [
    item[0] for item in lime_contributions
]

lime_weights = [
    item[1] for item in lime_contributions
]

lime_colours = [
    "green" if weight > 0 else "red"
    for weight in lime_weights
]

fig, ax = plt.subplots(figsize=(10, 7))

positions = np.arange(len(lime_weights))

ax.barh(
    positions,
    lime_weights,
    color=lime_colours
)

ax.set_yticks(positions)
ax.set_yticklabels(lime_labels)

ax.axvline(
    0,
    color="black",
    linewidth=0.8
)

ax.set_xlabel(
    "LIME contribution to HPV Positive prediction"
)

ax.set_title(
    "HECKTOR CNN – LIME Explanation (Epoch 50)"
)

ax.invert_yaxis()
plt.tight_layout()

lime_png = os.path.join(
    OUTPUT_DIR,
    "HECKTOR_CNN_Epoch50_LIME_Green_Red.png"
)

fig.savefig(
    lime_png,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# Save LIME values
pd.DataFrame(
    lime_contributions,
    columns=["Feature", "LIME_Contribution"]
).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_LIME_Values.csv"
    ),
    index=False
)

print("LIME plot saved:", lime_png)

# ------------------------------------------------------------
# 8. ROC CURVE POINTS
# ------------------------------------------------------------

print("\nCalculating ROC coordinates...")

y_true = np.asarray(y_test)

if y_true.ndim == 2:
    y_true = np.argmax(y_true, axis=1)
else:
    y_true = y_true.ravel()

y_probability = predict_proba_encoded(
    X_test_encoded
)[:, 1]

fpr, tpr, roc_thresholds = roc_curve(
    y_true,
    y_probability
)

roc_auc = roc_auc_score(
    y_true,
    y_probability
)

print("\n" + "=" * 65)
print("HECKTOR CNN — EPOCH 50 ROC POINTS")
print("=" * 65)

print(f"ROC-AUC = {roc_auc:.6f}")
print("\n(FPR, TPR)")

for x, y in zip(fpr, tpr):
    print(f"({x:.6f}, {y:.6f}),")

pd.DataFrame({
    "FPR": fpr,
    "TPR": tpr
}).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_ROC_Points.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# 9. PRECISION-RECALL CURVE POINTS
# ------------------------------------------------------------

precision, recall, pr_thresholds = precision_recall_curve(
    y_true,
    y_probability
)

pr_auc = average_precision_score(
    y_true,
    y_probability
)

print("\n" + "=" * 65)
print("HECKTOR CNN — EPOCH 50 PRECISION-RECALL POINTS")
print("=" * 65)

print(f"Average Precision = {pr_auc:.6f}")
print("\n(Recall, Precision)")

for x, y in zip(recall, precision):
    print(f"({x:.6f}, {y:.6f}),")

pd.DataFrame({
    "Recall": recall,
    "Precision": precision
}).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_PR_Points.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# 10. SAVE TEST PREDICTIONS
# ------------------------------------------------------------

pd.DataFrame({
    "y_true": y_true,
    "probability_class_1": y_probability
}).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_Test_Predictions.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# 11. CREATE ZIP FILE
# ------------------------------------------------------------

zip_path = (
    "/content/HECKTOR_CNN_Epoch50_SHAP_LIME_ROC_PR.zip"
)

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for filename in sorted(os.listdir(OUTPUT_DIR)):

        full_path = os.path.join(
            OUTPUT_DIR,
            filename
        )

        if os.path.isfile(full_path):
            zipf.write(
                full_path,
                arcname=filename
            )

print("\n" + "=" * 65)
print("RESULTS GENERATED SUCCESSFULLY")
print("=" * 65)

for filename in sorted(os.listdir(OUTPUT_DIR)):

    full_path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    print(
        filename,
        "-",
        round(os.path.getsize(full_path) / 1024, 2),
        "KB"
    )

print("\nZIP file:", zip_path)
print("Starting download...")

files.download(zip_path)
```

```python
# ============================================================
# HECKTOR CNN — EPOCH 50
# SHAP + LIME + ROC + PRECISION-RECALL
# COMPLETE CODE WITH DOWNLOADABLE RESULTS
# ============================================================

%pip -q install shap lime

import os
import zipfile
import numpy as np
import pandas as pd
import torch
import shap
import matplotlib.pyplot as plt

from lime.lime_tabular import LimeTabularExplainer

from sklearn.metrics import (
    roc_curve,
    precision_recall_curve,
    roc_auc_score,
    average_precision_score
)

from google.colab import files

# ============================================================
# 1. SETTINGS
# ============================================================

SEED = 42

np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

CHECKPOINT_PATH = (
    "/content/CNN_results_HECKTOR/checkpoints/"
    "cnn_best_val_roc_auc.pt"
)

OUTPUT_DIR = "/content/HECKTOR_CNN_EPOCH50_FINAL"

os.makedirs(OUTPUT_DIR, exist_ok=True)

feature_names = [
    "Age",
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

cat_cols = [
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status"
]

# ============================================================
# 2. LOAD EPOCH 50 CHECKPOINT
# ============================================================

try:
    checkpoint = torch.load(
        CHECKPOINT_PATH,
        map_location=device,
        weights_only=False
    )
except TypeError:
    checkpoint = torch.load(
        CHECKPOINT_PATH,
        map_location=device
    )

if isinstance(checkpoint, dict):

    if "model_state_dict" in checkpoint:
        state_dict = checkpoint["model_state_dict"]

    elif "state_dict" in checkpoint:
        state_dict = checkpoint["state_dict"]

    else:
        state_dict = checkpoint

else:
    state_dict = checkpoint

model.load_state_dict(state_dict)

model = model.to(device)
model.eval()

print("=" * 65)
print("HECKTOR CNN — EPOCH 50")
print("=" * 65)

print("Loaded checkpoint:", CHECKPOINT_PATH)
print("Selected checkpoint: best_val_roc_auc")
print("Checkpoint epoch: 50")

# ============================================================
# 3. PREPARE TRAINING AND TEST DATA
# ============================================================

X_train_original = X_train_df[feature_names].copy()
X_test_original = X_test_df[feature_names].copy()

X_train_encoded = preprocessor.transform(
    X_train_original
)

X_test_encoded = preprocessor.transform(
    X_test_original
)

if hasattr(X_train_encoded, "toarray"):
    X_train_encoded = X_train_encoded.toarray()

if hasattr(X_test_encoded, "toarray"):
    X_test_encoded = X_test_encoded.toarray()

X_train_encoded = np.asarray(
    X_train_encoded,
    dtype=np.float32
)

X_test_encoded = np.asarray(
    X_test_encoded,
    dtype=np.float32
)

print("\nTraining data:", X_train_encoded.shape)
print("Test data:", X_test_encoded.shape)

# ============================================================
# 4. CNN PREDICTION FUNCTION
# ============================================================

def predict_proba_encoded(X):

    X = np.asarray(
        X,
        dtype=np.float32
    )

    X_tensor = torch.tensor(
        X,
        dtype=torch.float32,
        device=device
    )

    model.eval()

    with torch.no_grad():

        # Do NOT use unsqueeze(1).
        # The CNN forward() method handles the channel dimension.

        logits = model(X_tensor)

        probabilities = torch.softmax(
            logits,
            dim=1
        )

    return probabilities.cpu().numpy()


# Check prediction function
check_predictions = predict_proba_encoded(
    X_test_encoded[:5]
)

print("\nPrediction function check:")
print("Prediction shape:", check_predictions.shape)
print(check_predictions)

assert check_predictions.shape == (5, 2)

print("CNN prediction function working correctly.")

# ============================================================
# 5. SHAP KERNEL EXPLAINER
# ============================================================

print("\n" + "=" * 65)
print("GENERATING SHAP EXPLANATIONS")
print("=" * 65)

rng = np.random.default_rng(SEED)

background_size = min(
    30,
    len(X_train_encoded)
)

explain_size = min(
    50,
    len(X_test_encoded)
)

background_indices = rng.choice(
    len(X_train_encoded),
    background_size,
    replace=False
)

background = X_train_encoded[
    background_indices
]

X_explain = X_test_encoded[
    :explain_size
]


def predict_class1(X):

    return predict_proba_encoded(X)[:, 1]


kernel_explainer = shap.KernelExplainer(
    predict_class1,
    background
)

raw_shap_values = kernel_explainer.shap_values(
    X_explain,
    nsamples=100,
    silent=False
)

shap_values = np.asarray(
    raw_shap_values
)

if shap_values.ndim == 3:

    shap_values = shap_values[:, :, 0]

shap_values = shap_values.reshape(
    len(X_explain),
    X_train_encoded.shape[1]
)

print(
    "Encoded SHAP values shape:",
    shap_values.shape
)

# ============================================================
# 6. GROUP SHAP VALUES INTO EIGHT ORIGINAL FEATURES
# ============================================================

encoded_feature_names = list(
    preprocessor.get_feature_names_out()
)

grouped_shap = np.zeros(
    (len(X_explain), len(feature_names))
)

grouped_columns = {}

for j, original_feature in enumerate(feature_names):

    matching_columns = []

    for i, encoded_name in enumerate(
        encoded_feature_names
    ):

        clean_name = encoded_name.split("__")[-1]

        if (
            clean_name == original_feature
            or clean_name.startswith(
                original_feature + "_"
            )
        ):

            matching_columns.append(i)

    if not matching_columns:

        matching_columns = [
            i
            for i, encoded_name in enumerate(
                encoded_feature_names
            )
            if original_feature in encoded_name
        ]

    if not matching_columns:

        raise ValueError(
            f"Could not identify encoded columns for "
            f"{original_feature}"
        )

    grouped_columns[original_feature] = matching_columns

    # Sum contributions from one-hot encoded columns
    grouped_shap[:, j] = shap_values[
        :, matching_columns
    ].sum(axis=1)


print("\nGrouped SHAP features:")

for feature, columns in grouped_columns.items():

    print(
        feature,
        ":",
        len(columns),
        "encoded column(s)"
    )

# ============================================================
# 7. PREPARE NUMERIC SHAP DISPLAY DATA
# ============================================================

X_display = X_test_original.iloc[
    :explain_size
].copy()

# Detect all categorical columns, including T-stage and N-stage
display_categorical_cols = [
    col for col in feature_names
    if (
        X_train_original[col].dtype == "object"
        or isinstance(
            X_train_original[col].dtype,
            pd.CategoricalDtype
        )
        or col in cat_cols
    )
]

display_category_mappings = {}

for col in display_categorical_cols:

    categories = list(
        pd.unique(
            X_train_original[col].dropna()
        )
    )

    mapping = {
        value: index
        for index, value in enumerate(categories)
    }

    display_category_mappings[col] = mapping

    X_display[col] = (
        X_display[col]
        .map(mapping)
        .fillna(-1)
        .astype(float)
    )

for col in feature_names:

    X_display[col] = pd.to_numeric(
        X_display[col],
        errors="raise"
    )

X_display = X_display[
    feature_names
].astype(float)

print(
    "\nSHAP display data shape:",
    X_display.shape
)

# ============================================================
# 8. GENERATE SHAP BEESWARM
# ============================================================

plt.figure(figsize=(10, 7))

shap.summary_plot(
    grouped_shap,
    X_display.to_numpy(),
    feature_names=feature_names,
    plot_type="dot",
    show=False
)

plt.title(
    "HECKTOR CNN – SHAP Beeswarm (Epoch 50)",
    fontsize=13
)

plt.tight_layout()

shap_png = os.path.join(
    OUTPUT_DIR,
    "HECKTOR_CNN_Epoch50_SHAP_Beeswarm.png"
)

plt.savefig(
    shap_png,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# Save SHAP values
pd.DataFrame(
    grouped_shap,
    columns=feature_names
).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_SHAP_Values.csv"
    ),
    index=False
)

print("\nSHAP beeswarm saved:", shap_png)

# ============================================================
# 9. LIME DATA PREPARATION
# ============================================================

print("\n" + "=" * 65)
print("GENERATING LIME EXPLANATION")
print("=" * 65)

X_train_lime = X_train_original.copy()
X_test_lime = X_test_original.copy()

categorical_indices = []
categorical_mappings = {}
categorical_names = {}

# Include all categorical features, including T-stage and N-stage
lime_categorical_cols = [
    col for col in feature_names
    if (
        X_train_original[col].dtype == "object"
        or isinstance(
            X_train_original[col].dtype,
            pd.CategoricalDtype
        )
        or col in cat_cols
    )
]

for col in lime_categorical_cols:

    idx = feature_names.index(col)

    categorical_indices.append(idx)

    categories = list(
        pd.unique(
            X_train_original[col].dropna()
        )
    )

    mapping = {
        value: i
        for i, value in enumerate(categories)
    }

    categorical_mappings[col] = mapping

    categorical_names[idx] = {
        i: str(value)
        for value, i in mapping.items()
    }

    X_train_lime[col] = (
        X_train_original[col]
        .map(mapping)
        .fillna(-1)
    )

    X_test_lime[col] = (
        X_test_original[col]
        .map(mapping)
        .fillna(-1)
    )

X_train_lime = X_train_lime.astype(float)
X_test_lime = X_test_lime.astype(float)

# ============================================================
# 10. LIME PREDICTION FUNCTION
# ============================================================

def lime_predict_proba(data):

    df = pd.DataFrame(
        data,
        columns=feature_names
    )

    for col in lime_categorical_cols:

        reverse_mapping = {
            code: value
            for value, code
            in categorical_mappings[col].items()
        }

        fallback_value = (
            X_train_original[col].mode().iloc[0]
        )

        df[col] = [
            reverse_mapping.get(
                int(round(value)),
                fallback_value
            )
            for value in df[col]
        ]

    encoded = preprocessor.transform(df)

    if hasattr(encoded, "toarray"):
        encoded = encoded.toarray()

    encoded = np.asarray(
        encoded,
        dtype=np.float32
    )

    return predict_proba_encoded(encoded)


# ============================================================
# 11. CREATE LIME EXPLAINER
# ============================================================

lime_explainer = LimeTabularExplainer(
    training_data=X_train_lime.to_numpy(),
    feature_names=feature_names,
    categorical_features=categorical_indices,
    categorical_names=categorical_names,
    mode="classification",
    class_names=[
        "HPV Negative",
        "HPV Positive"
    ],
    discretize_continuous=True,
    random_state=SEED
)

# Explain first test patient
patient_index = 0

lime_explanation = lime_explainer.explain_instance(
    data_row=X_test_lime.iloc[
        patient_index
    ].to_numpy(),
    predict_fn=lime_predict_proba,
    num_features=8,
    num_samples=3000
)

lime_contributions = lime_explanation.as_list(
    label=1
)

print(
    "\nLIME contributions for test patient:",
    patient_index
)

for feature, weight in lime_contributions:

    print(
        f"{feature}: {weight:.6f}"
    )

# ============================================================
# 12. LIME GREEN/RED PLOT
# ============================================================

lime_labels = [
    item[0]
    for item in lime_contributions
]

lime_weights = [
    item[1]
    for item in lime_contributions
]

lime_colours = [
    "green" if weight > 0 else "red"
    for weight in lime_weights
]

fig, ax = plt.subplots(
    figsize=(10, 7)
)

positions = np.arange(
    len(lime_weights)
)

ax.barh(
    positions,
    lime_weights,
    color=lime_colours
)

ax.set_yticks(positions)
ax.set_yticklabels(lime_labels)

ax.axvline(
    0,
    color="black",
    linewidth=0.8
)

ax.set_xlabel(
    "LIME contribution to HPV Positive prediction"
)

ax.set_title(
    "HECKTOR CNN – LIME Explanation (Epoch 50)"
)

ax.invert_yaxis()

plt.tight_layout()

lime_png = os.path.join(
    OUTPUT_DIR,
    "HECKTOR_CNN_Epoch50_LIME_Green_Red.png"
)

fig.savefig(
    lime_png,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# Save LIME values
pd.DataFrame(
    lime_contributions,
    columns=[
        "Feature",
        "LIME_Contribution"
    ]
).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_LIME_Values.csv"
    ),
    index=False
)

print("\nLIME graph saved:", lime_png)

# ============================================================
# 13. ROC CURVE POINTS
# ============================================================

print("\n" + "=" * 65)
print("HECKTOR CNN — EPOCH 50 ROC CURVE POINTS")
print("=" * 65)

y_true = np.asarray(y_test)

if y_true.ndim == 2:
    y_true = np.argmax(y_true, axis=1)
else:
    y_true = y_true.ravel()

y_probability = predict_proba_encoded(
    X_test_encoded
)[:, 1]

fpr, tpr, roc_thresholds = roc_curve(
    y_true,
    y_probability
)

roc_auc = roc_auc_score(
    y_true,
    y_probability
)

print(
    f"\nROC-AUC: {roc_auc:.6f}"
)

print("\n(FPR, TPR)")

for x, y in zip(fpr, tpr):

    print(
        f"({x:.6f}, {y:.6f}),"
    )

pd.DataFrame({
    "FPR": fpr,
    "TPR": tpr
}).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_ROC_Points.csv"
    ),
    index=False
)

# ============================================================
# 14. PRECISION-RECALL CURVE POINTS
# ============================================================

print("\n" + "=" * 65)
print("HECKTOR CNN — EPOCH 50 PRECISION-RECALL POINTS")
print("=" * 65)

precision, recall, pr_thresholds = (
    precision_recall_curve(
        y_true,
        y_probability
    )
)

pr_auc = average_precision_score(
    y_true,
    y_probability
)

print(
    f"\nAverage Precision: {pr_auc:.6f}"
)

print("\n(Recall, Precision)")

for x, y in zip(recall, precision):

    print(
        f"({x:.6f}, {y:.6f}),"
    )

pd.DataFrame({
    "Recall": recall,
    "Precision": precision
}).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_PR_Points.csv"
    ),
    index=False
)

# ============================================================
# 15. SAVE TEST PREDICTIONS
# ============================================================

pd.DataFrame({
    "y_true": y_true,
    "probability_class_1": y_probability
}).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_Test_Predictions.csv"
    ),
    index=False
)

# ============================================================
# 16. SAVE SUMMARY METRICS
# ============================================================

summary = pd.DataFrame({
    "Metric": [
        "ROC-AUC",
        "Average Precision (PR-AUC)",
        "Test observations",
        "SHAP observations explained",
        "SHAP background observations",
        "LIME patient index"
    ],
    "Value": [
        roc_auc,
        pr_auc,
        len(y_true),
        explain_size,
        background_size,
        patient_index
    ]
})

summary.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "HECKTOR_CNN_Epoch50_Summary_Metrics.csv"
    ),
    index=False
)

print("\nSummary metrics:")
print(summary.to_string(index=False))

# ============================================================
# 17. CREATE ZIP FILE
# ============================================================

zip_path = (
    "/content/HECKTOR_CNN_Epoch50_SHAP_LIME_ROC_PR.zip"
)

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for filename in sorted(
        os.listdir(OUTPUT_DIR)
    ):

        full_path = os.path.join(
            OUTPUT_DIR,
            filename
        )

        if os.path.isfile(full_path):

            zipf.write(
                full_path,
                arcname=filename
            )

# ============================================================
# 18. VERIFY OUTPUTS
# ============================================================

print("\n" + "=" * 65)
print("GENERATED FILES")
print("=" * 65)

for filename in sorted(
    os.listdir(OUTPUT_DIR)
):

    full_path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    print(
        filename,
        "-",
        round(
            os.path.getsize(full_path) / 1024,
            2
        ),
        "KB"
    )

print("\nZIP file:", zip_path)

# ============================================================
# 19. DOWNLOAD ALL RESULTS
# ============================================================

print("\nDownloading all results...")

files.download(zip_path)
```
